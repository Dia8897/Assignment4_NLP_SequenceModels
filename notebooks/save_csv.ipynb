# Import

```python
from pathlib import Path
import pandas as pd
import numpy as np
import os
```

# Path setup

```python
PROJECT_ROOT=Path.cwd().parent
DATA_DIR=PROJECT_ROOT/"data"/"aclImdb"
TRAIN_DIR=DATA_DIR/"train"
TEST_DIR=DATA_DIR/"test"

print("Dataset exists:", DATA_DIR.exists())
print("Training folder exists:", TRAIN_DIR.exists())
print("Testing folder exists:", TEST_DIR.exists())
```

# Loading reviews

## Helper function

```python
def load_reviews(folder_path,label):
    reviews=[]
    for file_path in folder_path.glob("*.txt"):
        with open(file_path,"r",encoding="utf-8") as file:
            text=file.read()
            reviews.append({
                "text":text,
                "label":label
            })
    return pd.DataFrame(reviews)
```

## loading

```python
train_pos = load_reviews(TRAIN_DIR / "pos", 1)
train_neg = load_reviews(TRAIN_DIR / "neg", 0)

test_pos = load_reviews(TEST_DIR / "pos", 1)
test_neg = load_reviews(TEST_DIR / "neg", 0)
```

## Concatenate 

```python
train_df = pd.concat([train_pos, train_neg], ignore_index=True)
test_df = pd.concat([test_pos, test_neg], ignore_index=True)
```

```python
print("Training set shape:", train_df.shape)
print("Testing set shape:", test_df.shape)
train_df.head()
```

## Save csv

```python
train_df.to_csv(DATA_DIR / "train.csv", index=False)
test_df.to_csv(DATA_DIR / "test.csv", index=False)
print("Datasets saved successfully")
```
